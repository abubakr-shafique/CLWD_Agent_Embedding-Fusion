```python
import os
import openslide
import project_dirs as pdir
```

```python
# Desired local directory
WSI_dir = pdir.DATA_DIR

extensions = ['svs'] ## File extensions you want to search in the directory
WSI_List = [fn for fn in os.listdir(WSI_dir) if any(fn.endswith(ext) for ext in extensions)] ##Gets all the files in the directory with the above mentioned extension
```

```python
def print_wsi_info(wsi_path):

    slide = openslide.OpenSlide(wsi_path)

    print("=" * 80)
    print(f"WSI: {os.path.basename(wsi_path)}")
    print("=" * 80)

    # ---------------------------------------------------------
    # Basic information
    # ---------------------------------------------------------

    print("\n--- Basic Information ---")

    print(f"Vendor              : {slide.properties.get(openslide.PROPERTY_NAME_VENDOR)}")
    print(f"Level count         : {slide.level_count}")

    # Level 0 dimensions
    width, height = slide.dimensions

    print(f"Level 0 dimensions  : {width} x {height}")
    print(f"Level 0 width       : {width}")
    print(f"Level 0 height      : {height}")

    # ---------------------------------------------------------
    # MPP
    # ---------------------------------------------------------

    print("\n--- Resolution ---")

    mpp_x = slide.properties.get(openslide.PROPERTY_NAME_MPP_X)
    mpp_y = slide.properties.get(openslide.PROPERTY_NAME_MPP_Y)

    print(f"MPP X               : {mpp_x} µm/pixel")
    print(f"MPP Y               : {mpp_y} µm/pixel")

    # ---------------------------------------------------------
    # Objective power
    # ---------------------------------------------------------

    objective_power = slide.properties.get(
        openslide.PROPERTY_NAME_OBJECTIVE_POWER
    )

    print(f"Objective power     : {objective_power}x")

    # ---------------------------------------------------------
    # All pyramid levels
    # ---------------------------------------------------------

    print("\n--- Pyramid Levels ---")

    print(
        f"{'Level':<10}"
        f"{'Width':<15}"
        f"{'Height':<15}"
        f"{'Downsample':<15}"
    )

    print("-" * 55)

    for level in range(slide.level_count):

        width, height = slide.level_dimensions[level]
        downsample = slide.level_downsamples[level]

        print(
            f"{level:<10}"
            f"{width:<15}"
            f"{height:<15}"
            f"{downsample:<15.2f}"
        )

    # ---------------------------------------------------------
    # Associated images
    # ---------------------------------------------------------

    print("\n--- Associated Images ---")

    if slide.associated_images:
        for name, image in slide.associated_images.items():
            print(f"{name:<20}: {image.size}")
    else:
        print("None")


    print("=" * 80)

    slide.close()
```

```python
for W in WSI_List:
    print_wsi_info(os.path.join(WSI_dir, W))
    break
```
